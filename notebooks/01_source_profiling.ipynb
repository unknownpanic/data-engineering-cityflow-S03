```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent

customers = pd.read_csv(root / "data" / "raw" / "customers.csv")
merchants = pd.read_csv(root / "data" / "raw" / "merchants.csv")
products = pd.read_csv(root / "data" / "raw" / "products.csv")
orders = pd.read_csv(root / "data" / "raw" / "orders.csv")

datasets = {
    "Customers": customers,
    "Merchants": merchants,
    "Products": products,
    "Orders": orders
}

for name, df in datasets.items():
    print(f"\n{'='*40}")
    print(f"Профілювання датасету: {name}")
    print(f"{'='*40}")
    
    print("\n1. Форма (Shape):", df.shape)
    
    print("\n2. Перші 3 рядки (Head):")
    display(df.head(3))
    
    print("\n3. Останні 3 рядки (Tail):")
    display(df.tail(3))
    
    print("\n4. Інформація (Info):")
    df.info()
    
    print("\n5. Дублікати:", df.duplicated().sum())
    
    print("\n6. Пропущені значення (Missing):")
    print(df.isna().sum()[df.isna().sum() > 0])
    
    print("\n7. Опис числових даних (Describe):")
    display(df.describe())
    
print("\nДіапазони дат:")
print("Customers registered_at:", customers['registered_at'].min(), "-", customers['registered_at'].max())
print("Merchants opened_at:", merchants['opened_at'].min(), "-", merchants['opened_at'].max())
print("Products updated_at:", products['updated_at'].min(), "-", products['updated_at'].max())
print("Orders created_at:", orders['created_at'].min(), "-", orders['created_at'].max())
```

### Висновки щодо структури даних:
* **customers.csv**: Один рядок — це один унікальний клієнт. Кандидатний ключ — `customer_id`.
* **merchants.csv**: Один рядок — це один заклад-партнер. Кандидатний ключ — `merchant_id`.
* **products.csv**: Один рядок — це один товар у конкретному закладі. Кандидатний ключ — `product_id`. Зовнішній ключ — `merchant_id`.
* **orders.csv**: Один рядок — це одне замовлення клієнта в конкретному закладі. Кандидатний ключ — `order_id`. Зовнішні ключі: `customer_id` та `merchant_id`.

```python
plt.figure(figsize=(10, 5))
orders['order_status'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Кількість замовлень за статусами')
plt.xlabel('Статус')
plt.ylabel('Кількість')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))
orders['created_at_parsed'] = pd.to_datetime(orders['created_at'], errors='coerce')
orders_by_date = orders['created_at_parsed'].dt.date.value_counts().sort_index()

orders_by_date.plot(kind='line', marker='o', color='coral')
plt.title('Динаміка замовлень за датою')
plt.xlabel('Дата')
plt.ylabel('Кількість замовлень')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
